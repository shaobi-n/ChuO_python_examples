```python
import requests   
from dotenv import load_dotenv
import os 
import json
load_dotenv(dotenv_path= '.env')
```

```python
#Load .env variables 
api_key = os.getenv('qualtrics_api_key')
data_center_id = os.getenv('data_center_id')
client_id = os.getenv('client_id')
client_secret = os.getenv('client_secret')
```

```python
#OAuth 
auth_url = f"https://{data_center_id}.qualtrics.com/oauth2/token"

#Defining POST request credentials 
data = {
  'grant_type': 'client_credentials',
  'client_id': client_id,
  'client_secret': client_secret,
  'scope': 'read:surveys read:survey_responses'
}

#Make POST request to AUTH url
response = requests.post(auth_url, data = data)

if response.status_code == 200:
  token_info = response.json()
  access_token = token_info['access_token']
  print(access_token)
```

```python
#Define survey_url
survey_url = f"https://{data_center_id}.qualtrics.com/API/v3/surveys"

#Headers for HTTP GET request
headers = {
  'Authorization': f'Bearer {access_token}',
  'Content-Type': 'application/json'
}

#Make GET request
response = requests.get(url = survey_url, headers = headers)
surveys = response.json()
print(surveys)
```

```python
#Retrieve survey ID's 
for survey in surveys['result']['elements']:
  survey_id = survey['id']
  survey_name = survey['name']
  print(f'Survey Name: {survey_name}, Survey ID: {survey_id}')
```
